```python
# to create a new column 'height_category' based on the values in the 'height_cm' column.
bios_new['height_category'] = bios_new['height_cm'].apply(
    lambda x: 'Short' if x<165 else 'Average' if x<185 else 'Tall'
    )
bios_new.head()
```

```python
# to create 'height_category' using numpy.select() (recommended for multiple conditions)
import numpy as np

conditions = [
    bios_new['height_cm'] < 165,
    bios_new['height_cm'] < 185
]

choices = ['Short', 'Average']

bios_new['height_category'] = np.select(
    conditions,
    choices,
    default='Tall'
)
bios_new.head()
```

```python
# to create a new column 'weight_category' based on the values in the 'weight_kg' column. 
# We will use the apply method to apply a function that categorizes the weight into 'Lightweight', 'Middleweight' and 'Heavyweight'. 
# axis=1 is used to apply the function to each row of the dataframe. 
def categorize_weight(row):
    if row['height_cm'] < 165 and row['weight_kg'] < 70:
        return 'Lightweight'
    elif row['height_cm'] < 185 and row['weight_kg'] < 85:
        return 'Middleweight'
    else:
        return 'Heavyweight'
    
bios_new['weight_category'] = bios_new.apply(categorize_weight, axis=1)
bios_new.head()
```

### Merging and Concatenating Data

```python
nocs = pd.read_csv('../data/external/noc_regions.csv')
bios = pd.read_csv('../data/external/bios.csv')
```

```python
# to merge the bios and nocs dataframes on the born_country and NOC columns respectively. 
# We will use a left join to keep all the rows from the bios dataframe and add the corresponding region information from the nocs dataframe.
# how='left' mean Keep all rows from bios, even if no match is found in nocs
bios_new = pd.merge(bios, nocs, left_on='born_country', right_on='NOC', how='left')
bios_new.head()
```

```python
# for identifying invalid country codes in bios_new
bios_nan = bios_new[bios_new['NOC_x'].isna()]
bios_nan.head()
```

```python
# to select rows where the NOC_x column (which is the born_country from bios) is not equal to the region column (which is the region from nocs).
# One practical note: if either NOC_x or region contains NaN, then != treats NaN != NaN as True. 
# That means rows with missing values may appear in your result unless you handle missing data explicitly.
bios_new.loc[bios_new['NOC_x'] != bios_new['region']][['name', 'born_country', 'NOC_x', 'region']]
```

```python
# to select rows where the born_country column is either 'USA' or 'GBR', 
# and then concatenate the resulting dataframes for USA and GBR into a new dataframe called new_df. Concat puts one DF on top of another DF.
usa = bios_new.loc[bios_new['born_country'] == 'USA']
gbr = bios_new.loc[bios_new['born_country'] == 'GBR']

new_df = pd.concat([usa, gbr], ignore_index=True)
```

```python
# to display the last 2 rows of the new_df dataframe with USA and GBR athletes.
new_df.tail(2)
```

```python
# Merging two DF on common ID
result = pd.read_csv('../data/external/results.csv')
combined_df = pd.merge(result, new_df, on='athlete_id', how='left')
combined_df.head()
```

### Handling Null Values


```python
# modify existing DF to insert some null values and then handle them
people = pd.read_csv('../data/external/some-people.csv')
people.head()
```

```python
# to fill null values in 'age' column with the mean of the 'age' column. This will modify the original dataframe.
people = people.fillna(people['age'].mean())
people.head()
```

```python
# to fill null values with linear interpolation method. This will modify the original dataframe.
people.loc[3:4, 'age'] = np.nan # first we are inserting null values to row 3 and 4 and column 'age' to demonstrate interpolation method
people['age'] = people['age'].interpolate() # interpolation method fills the null values by calculating the average of the previous and next non-null values in the 'age' column.
people.head(6)
```

```python
# dropna() method is used to remove rows with null values in the 'age' column. 
# subset parameter specifies the column to check for null values. inplace=True modifies the original dataframe.
people.loc[1:2, 'age'] = np.nan
people.dropna(subset=['age'], inplace=True)
people.head(5)
```

```python
# to calculate the average height in centimeters for each unique value in the 'NOC' column. 
# This will return a new Series with the NOC as the index and the average height as the values.
bios.groupby(['NOC'])['height_cm'].mean()
```

```python
# `agg()` method allows us to perform multiple aggregation operations on different columns of the grouped data.
# `agg()` method takes a dictionary where the keys are the column names and the values are the aggregation functions to be applied to those columns.
# Here, we are grouping the data by both 'NOC' and 'born_city' columns, and then applying the mean aggregation function to both 'height_cm' and 'weight_kg' columns.
# Note: Since there are some null values in 'height_cm' and 'weight_kg' columns, 
# we are using interpolate() method to fill those null values before applying the groupby and agg operations.
bios['height_cm'] = bios['height_cm'].interpolate()
bios['weight_kg'] = bios['weight_kg'].interpolate()
bios.groupby(['NOC', 'born_city']).agg({'height_cm': 'mean', 'weight_kg': 'mean'}).head()
```

```python
coffee = pd.read_csv('../data/external/coffee.csv')
coffee.head()
```

```python
# to create a pivot table that shows the total units sold for each coffee type on each day.
# The pivot() method is used to reshape the data in the coffee dataframe.
# The columns parameter specifies the column to use as the new columns in the pivot table,
# the index parameter specifies the column to use as the new index in the pivot table, and
# the values parameter specifies the column to use for populating the values in the pivot table.
# After creating the pivot table, we are reindexing it to ensure that the days of the week are in the correct order.
pivot = coffee.pivot(
    columns='Coffee Type', 
    index='Day', 
    values='Units Sold'
    )
pivot = pivot.reindex([
    'Monday',
    'Tuesday',
    'Wednesday',
    'Thursday',
    'Friday',
    'Saturday',
    'Sunday'
])
pivot
```

```python
# to access the total units sold for 'Latte' on 'Monday' from the pivot table. This will return a single value which is the total units sold for 'Latte' on 'Monday'.
x = pivot.loc['Monday','Latte'] #np.int64(15)
int(x)
```

```python
# to calculate the total units sold for each coffee type across all days. This will return a Series with the coffee types as the index and the total units sold as the values.
pivot.sum()
```

```python
# to calculate the total units sold for each day across all coffee types. This will return a Series with the days of the week as the index and the total units sold as the values.
pivot.sum(axis=1)
```

```python
# to convert the 'born_date' column to a datetime object and then group the data by the year of birth and count the number of athletes born in each year. 
# Finally, we are sorting the result by the count of athletes in descending order.
# `reset_index()` is used to convert the Series returned by groupby into a DataFrame, and `sort_values()` is used to sort the DataFrame by the 'athlete_id' column in descending order.
bios['born_date'] = pd.to_datetime(bios['born_date'])
bios.groupby(bios['born_date'].dt.year)['athlete_id'].count().reset_index().sort_values('athlete_id', ascending=False)
```

```python
# to create a new column 'latte_revenue_by_day' that contains the total revenue for 'Latte' on each day.
# We are filtering the coffee dataframe to include only rows where the 'Coffee Type' is 'Latte', then grouping the data by 'Day' and summing the 'Revenue' for each day.
# After calculating the total revenue for 'Latte' on each day, we are finding the day with the highest revenue and the corresponding revenue amount.
# `idxmax()` is used to get the index of the maximum value in the Series, which in this case will be the day with the highest revenue for 'Latte'.
# `max()` is used to get the maximum value in the Series, which in this case will be the highest revenue for 'Latte'.
latte_revenue_by_day = coffee.loc[coffee['Coffee Type'] == 'Latte'].groupby('Day')['Revenue'].sum()
best_day = latte_revenue_by_day.idxmax()
best_revenue = latte_revenue_by_day.max()

latte_revenue_by_day, best_day, best_revenue
```

### Advanced Functionality

```python
# to create a new column 'Price' in the coffee dataframe by mapping the 'Coffee Type' column to a price using a dictionary.
# The map() method is used to map the values in the 'Coffee Type' column to the corresponding prices in the price_map dictionary.
# We are creating a price_map dictionary where the keys are the coffee types and the values are their corresponding prices.
# After creating the 'Price' column, we are calculating the 'Revenue' by multiplying the 'Units Sold' column with the 'Price' column.
price_map = {
    'Espresso': 3.99,
    'Latte': 5.99
}

coffee['Price'] = coffee['Coffee Type'].map(price_map)
coffee['Revenue'] = coffee['Units Sold'] * coffee['Price']
coffee.head()
```

```python
# to create a new column 'yesterday_revenue' that contains the revenue from two days ago.
# The shift() method is used to shift the values in the 'Revenue' column down by 2 rows, 
# which means that the value in the 'yesterday_revenue' column for a given row will be the value from the 'Revenue' column that is 2 rows above it.
# shift(2) is done cause there are two types of coffee and we want to get the revenue from the same coffee type from two days ago.
coffee['yesterday_revenue'] = coffee['Revenue'].shift(2)
coffee.head(8)
```

```python
# to calculate the percentage change in revenue compared to two days ago and store it in a new column 'pct_change'.
# We are calculating the percentage change in revenue by dividing the current revenue by the yesterday's revenue, 
# subtracting 1 from the result to get the change, and then multiplying by 100 to convert it to a percentage.
coffee['pct_change'] = (coffee['Revenue'] / coffee['yesterday_revenue'] - 1) * 100
coffee.head(8)
```

```python
# to rank the athletes based on their height in centimeters and create a new column 'height_rank' that contains the rank of each athlete.
# The rank() method is used to assign a rank to each value in the 'height_cm
# The two athletes with height 223 occupy positions 2 and 3. Their rank is: (2 + 3) / 2 = 2.5
bios['height_rank'] = bios['height_cm'].rank(ascending=False)
bios.sort_values(['height_rank'])[['name', 'height_cm', 'height_rank']].head(10)
```

```python
# to calculate the cumulative sum of the 'Revenue' column and store it in a new column 'Cumulative_Revenue'.
# The cumsum() method is used to calculate the cumulative sum of the 'Revenue' column, which means that each value in the 'Cumulative_Revenue' 
# column will be the sum of all the previous values in the 'Revenue' column up to that point.
coffee['Cumulative_Revenue'] = coffee['Revenue'].cumsum()
coffee.head()
```

```python
latte = coffee.loc[coffee['Coffee Type'] == 'Latte'].copy()
latte.head()
```

```python
# to calculate the rolling sum of the 'Units Sold' column for a window of 3 days and store it in a new column '3_day_cumulative_units'.
# The rolling() method is used to calculate the rolling sum of the 'Units Sold' column for a window of 3 days, 
# which means that each value in the '3_day_cumulative_units' column will be the sum of the current and previous 2 values in the 'Units Sold' column.
latte['3_day_cumulative_units'] = latte['Units Sold'].rolling(3).sum()
latte
```
